```python
print("Avant :", len(df))

df_nettoye = df.drop_duplicates(subset=df.columns.drop("Row ID"))

print("Après :", len(df_nettoye))
print("Lignes supprimées :", len(df) - len(df_nettoye))
```

```python
# 1. AVANT : vérifier le type
print("Avant :", df_nettoye["Order Date"].dtype)

# 2. CONVERTIR le texte en date
df_nettoye["Order Date"] = pd.to_datetime(df_nettoye["Order Date"], format="%m/%d/%Y")
df_nettoye["Ship Date"] = pd.to_datetime(df_nettoye["Ship Date"], format="%m/%d/%Y")

# 3. APRÈS : vérifier que ça a marché
print("Après :", df_nettoye["Order Date"].dtype)
print("Après :", df_nettoye["Ship Date"].dtype)
```

```python
df_nettoye["Annee"] = df_nettoye["Order Date"].dt.year
df_nettoye["Mois"] = df_nettoye["Order Date"].dt.month
df_nettoye["Delai_livraison"] = (df_nettoye["Ship Date"] - df_nettoye["Order Date"]).dt.days

df_nettoye[["Order Date", "Ship Date", "Annee", "Mois", "Delai_livraison"]].head()
```

```python
df_nettoye["Delai_livraison"] = (df_nettoye["Ship Date"] - df_nettoye["Order Date"]).dt.days
```

```python
#Bonus : et si on regardait le délai moyen par mode de livraison
df_nettoye.groupby("Ship Mode")["Delai_livraison"].mean()
```

```python
df_nettoye["Marge"] = df_nettoye["Profit"] /df_nettoye["Sales"]* 100
pourcentage
df_nettoye[["Sales", "Profit", "Marge"]].head()
```
